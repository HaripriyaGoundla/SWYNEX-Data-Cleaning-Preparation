```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_csv(list(uploaded.keys())[0])

print("Dataset loaded successfully!")

display(df)
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df.replace(
    ["", " ", "N/A", "NA", "NULL", "null", "-"],
    np.nan,
    inplace=True
)

print(df.isnull().sum())
```

```python
df[df.isnull().any(axis=1)]
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print("Duplicate emails:")
display(df[df.duplicated(subset=["email"], keep=False)])
```

```python
df = df.drop_duplicates(subset=["email"], keep="first")

print("Rows after removing duplicates:", df.shape[0])
```

```python
df["Full Name"] = df["Full Name"].str.strip()

df["Full Name"] = df["Full Name"].str.title()

display(df[["id", "Full Name"]])
```

```python
df["email"] = df["email"].str.strip().str.lower()

display(df[["id", "email"]])
```

```python
df["country"] = df["country"].str.strip().str.title()

df["country"] = df["country"].replace({
    "Usa": "United States",
    "U.S.A.": "United States",
    "Uk": "United Kingdom",
    "De": "Germany"
})

display(df[["id", "country"]])
```

```python
df["is_active"] = df["is_active"].str.strip().str.lower()

df["is_active"] = df["is_active"].replace({
    "yes": True,
    "y": True,
    "true": True,
    "1": True,
    "active": True,
    "no": False,
    "n": False,
    "false": False,
    "0": False,
    "inactive": False
})

display(df[["id", "is_active"]])
```

```python
df["amount"] = (
    df["amount"]
    .astype(str)
    .str.strip()
    .str.replace("$", "", regex=False)
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["amount"] = pd.to_numeric(
    df["amount"],
    errors="coerce"
)

display(df[["id", "amount"]])
```

```python
df.loc[df["id"] == 9, "amount"] = 1250.00

display(df[["id", "amount"]])
```

```python
df["negative_amount"] = df["amount"] < 0

display(df[["id", "amount", "negative_amount"]])
```

```python
df["signup date"] = pd.to_datetime(
    df["signup date"],
    errors="coerce",
    format="mixed"
)

display(df[["id", "signup date"]])
```

```python
display(df)
```

```python
print("Duplicate customer emails:")

display(
    df[df.duplicated(subset=["email"], keep=False)]
    [["id", "Full Name", "email"]]
)
```

```python
df = df.drop_duplicates(
    subset=["email"],
    keep="first"
)

print("Rows after removing duplicate customers:", df.shape[0])
```

```python
df = df[df["id"] != 13]

print("Rows after removing invalid record:", df.shape[0])
```

```python
df.info()
```

```python
df["is_active"] = df["is_active"].astype("boolean")

print(df["is_active"].dtype)
```

```python
df.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
